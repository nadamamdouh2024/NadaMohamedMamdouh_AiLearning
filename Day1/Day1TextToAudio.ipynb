```python
import PyPDF2
```

```python
pdf=open("/content/aiparagraph.pdf","rb")    #rb=> read binary
```

```python
#pdf reader

pdfReader=PyPDF2.PdfReader(pdf)
countpages = len(pdfReader.pages)
textlist=[]
```

```python
for i in range(countpages):
  try:
    page = pdfReader.pages[i]
    text = page.extract_text()
    textlist.append(text)

  except :
    pass
```

```python
textstring="".join(textlist).strip()
print(textstring)
```

```python
language= 'en'
```

```python
audio = gTTS(text=textstring,lang=language,slow=False)
```

```python
#save audio

audio.save("TtoS.mp3")
```
